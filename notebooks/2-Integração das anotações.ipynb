# **Annotation integration**

```python
import pandas as pd
```

## **/nh/ sample**

```python
annotator_0 = pd.read_csv("../data/processed/annotator-0_nasal-palatal_stratified_sample.csv")
annotator_1 = pd.read_csv("../data/processed/annotator-1_nasal-palatal_stratified_sample.csv")
annotator_2 = pd.read_csv("../data/processed/annotator-2_nasal-palatal_stratified_sample.csv")
```

```python
nasal_palatal = annotator_0[[col for col in annotator_0.columns if not col.startswith("realizacao_") and not col.startswith("numero_de_vezes_")]].copy()
nasal_palatal.info()
```

```python
for df in [annotator_0, annotator_1, annotator_2]:
    for col in df.columns:
        if col.startswith("realizacao_") or col.startswith("numero_de_vezes_"):
            nasal_palatal[col] = df[col]
            nasal_palatal[col] = nasal_palatal[col].astype("int", errors="ignore")

nasal_palatal.info()
```

```python
nasal_palatal.to_csv("../data/finalized/annotated_nasal-palatal_stratified_sample.csv", index=False)
```
